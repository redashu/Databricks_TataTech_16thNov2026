```python
# using volume data to read using spark read function 
file_path = "/Volumes/delvex_workspace/default/ashu-tata/student_performance_dataset.csv"
#df = spark.read.csv(file_path, header=True, inferSchema=True)
df = spark.read.option("header", "true").option("inferSchema", "true").csv(file_path)
display(df)
```

## Inserting data in databricks table -- save -- delta format


```python
# lets save as table 
df.write.format("delta").mode("overwrite").saveAsTable("student_performance")
```
