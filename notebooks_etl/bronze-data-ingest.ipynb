```python
%sql
CREATE OR REPLACE TABLE
    databricks_ukwest.ashu_etl.ashu_student_perform_bronze
USING DELTA
AS
SELECT
    *,
    current_timestamp() AS bronze_ingested_at
FROM read_files(
    '/Volumes/databricks_ukwest/ashu_etl/ashu_etl_volume/student_performance_dataset.csv',
    format => 'csv',
    header => true,
    inferColumnTypes => false,
    nullValue => 'None'
);
```
